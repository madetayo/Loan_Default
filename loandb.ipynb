```python
bins = [520, 600, 650, 700, 750, np.inf]
labels = ['520-599', '600-649', '650-699', '700-749', '750+']

borrower_profiles['credit_band'] = pd.cut(borrower_profiles['credit_score'], bins=bins, labels=labels, right=False)
```

```python
bins = [0, 21, 41, 61, 81, np.inf]
labels = ['0-20.99', '21-40.99', '41-60.99', '61-80.99', '81+']

loan_applications['dti_band'] = pd.cut(loan_applications['dti_ratio'], bins=bins, labels=labels, right=False)
```

```python
bins = [18, 31, 44, 57, np.inf]
labels = ['18-30', '31-43', '44-56', '57+']

borrower_profiles['age_group'] = pd.cut(borrower_profiles['age'], bins=bins, labels=labels, right=False)
```

```python
bins = [5, 8, 11, 14, np.inf]
labels = ['5-7.99', '8-10.99', '11-13.99', '14+']

loan_applications['interest_rate'].rename = 'interest_rate_band'
```

```python
from sqlalchemy import create_engine
import urllib.parse

connection_string = urllib.parse.quote_plus(
    'DRIVER={ODBC Driver 17 for SQL Server};'
    'SERVER=AyoDAnalyst\\SQLEXPRESS01;'
    'DATABASE=LoanDB;'
    'Trusted_Connection=yes;'
)

engine = create_engine(
    f'mssql+pyodbc:///?odbc_connect={connection_string}'
)
```

```python
borrower_profiles.to_sql(
    'borrower_profiles',
    engine,
    schema='dbo',
    if_exists='replace',
    index=False
)

loan_applications.to_sql(
    'loan_applications',
    engine,
    schema='dbo',
    if_exists='replace',
    index=False
)
```
